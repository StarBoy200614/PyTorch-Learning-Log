##Reshaping, stacking, squeezing and unsqueezing

```python
# Create a tensor
import torch
x = torch.arange(1., 10.)
x, x.shape
```

```python
# Add an extra dimension
x_reshaped = x.reshape(1, 9)
x_reshaped, x_reshaped.shape
```

```python
# Change the view
z = x.view(1, 9)
z, z.shape
```

```python
#Changing z changes x because the view of tensor shares the same memory as the original tensor
z[:, 0] = 5
z, x
```

```python
# Stack tensors on top
x_stacked = torch.stack([x_reshaped,x_reshaped,x_reshaped,x_reshaped], dim = 0)
x_stacked
```

```python
# torch.squeeze() -  removes all single dimensions from a target tensor
x_reshaped, x_reshaped.squeeze()
x_squeezed = x_reshaped.squeeze()
```

```python
x_reshaped.shape, x_reshaped.squeeze().shape, x.shape
```

```python
# torch.unsqueeze() - adds a single dmension to a tagret tensor at a specific dim (dimension)

print(f"Previous target: {x_squeezed}")
print(f"Previous shape: {x_squeezed.shape}")

x_unsqueezed = x_squeezed.unsqueeze(dim=0)

print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
```

```python
# torch.permute - rearranges the dimensions of a target tensor in a specified order
x_original = torch.rand(size=(224, 244, 3)) # [height, width, color_channels]

#Permute the originak tensor to reaarange the axis (or dim) order
x_permuted = x_original.permute(2, 0, 1) #shifts axis 0->1, 1->2, 2->0
print(f"Previous shape: {x_original.shape}")
print(f"New shape: {x_permuted.shape}") # [colors_channels, height, widht]
```

```python
x_original[0,0,0] = 2345.
x_permuted[0,0,0] #Shallow copy of original
```

##PyTorch tensors & NumPy

* Data in NumPy, want in PyTorch tensor -> torch.from_numpy(ndarray)
* PyTorch tensor -> NumPy -> torch.tensor.numpy()

```python
# NumPy array to tensor
import torch
import numpy as np

array = np.arange(1.,8.)
tensor = torch.from_numpy(array).type(torch.float32) #when converting from numpy to pytorch, it reflects the default datatype of numpy array
array, tensor
```

```python
array.dtype, tensor.dtype
```

```python
#changing the value of array and ovserving tensor
array = array + 1
array ,tensor
```

```python
# Tensor to Numpy array
tensor = torch.ones(7)
numpy_tensor = tensor.numpy()
tensor, numpy_tensor
```

```python
numpy_tensor.dtype
```

```python
#Changing tensor and observing numpy_tensor
tensor = tensor + 1
numpy_tensor, tensor
```

##Reproducibility (trying to take the random out of random)

In short how a nueral network learns is:

'Starrt with random numbers -> tensor operations -> update random numbers to try and make them better representations of the data -> again -> again

To reduce the randomness in neural network we introduce concept of a random seed.


```python
import torch

# Create two random tensors
rand_a = torch.rand(3,4)
rand_b = torch.rand(3,4)
print(rand_a)
print(rand_b)
print(rand_a == rand_b)
```

```python
# Let's make some random but reproducable tensors

import torch

#Set the random seed
RANDOM_SEED = 7
torch.manual_seed(RANDOM_SEED)
rand_c = torch.rand(3,4)

torch.manual_seed(RANDOM_SEED)
rand_d = torch.rand(3,4)

print(rand_c)
print(rand_d)
print(rand_c == rand_d)
```
